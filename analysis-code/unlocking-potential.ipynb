```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentPerformanceFactors.csv')

grouped_ld = df.groupby(['Learning_Disabilities', 'Internet_Access'])['Exam_Score'].agg(['mean', 'count']).reset_index()

plt.figure(figsize=(8.5, 5.2), dpi=300)

palette = {'No': '#E76F51', 'Yes': '#2A9D8F'}

ax = sns.barplot(
    data=grouped_ld,
    x='Learning_Disabilities',
    y='mean',
    hue='Internet_Access',
    palette=palette,
    edgecolor='white',
    linewidth=1.2,
    width=0.45
)

plt.title("Unlocking Potential: Learning Differences & Digital Inclusion", fontsize=13, fontweight='bold', pad=15, color='#1D3557')
plt.xlabel("Presence of Learning Disability", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylabel("Average Exam Score", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylim(60, 72)

for p in ax.patches:
    height = p.get_height()
    if not pd.isna(height) and height > 0:
        ax.annotate(f'{height:.1f}',
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom',
                    fontsize=10, color='#1D3557', fontweight='bold',
                    xytext=(0, 3),
                    textcoords='offset points')

plt.legend(title="Internet Access", frameon=True, facecolor='white', edgecolor='none')
plt.grid(True, linestyle='--', alpha=0.3, axis='y')
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
