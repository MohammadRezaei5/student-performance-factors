```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentPerformanceFactors.csv')

df['Attendance_Bin'] = pd.cut(df['Attendance'], bins=[50, 60, 70, 80, 90, 100], labels=['50-60%', '61-70%', '71-80%', '81-90%', '91-100%'])

grouped_attendance = df.groupby(['Attendance_Bin', 'Peer_Influence'], observed=False)['Exam_Score'].mean().reset_index()

plt.figure(figsize=(10, 5.5), dpi=300)

palette = {'Positive': '#1D3557', 'Neutral': '#457B9D', 'Negative': '#E63946'}

ax = sns.lineplot(
    data=grouped_attendance,
    x='Attendance_Bin',
    y='Exam_Score',
    hue='Peer_Influence',
    palette=palette,
    marker='o',
    markersize=8,
    linewidth=2.5
)

plt.title("The Social Compass: Attendance, Peer Influence, and Academic Trajectory", fontsize=14, fontweight='bold', pad=15, color='#1D3557')
plt.xlabel("Class Attendance Rate", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylabel("Average Exam Score", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylim(60, 75)

plt.legend(title="Peer Influence", frameon=True, facecolor='white', edgecolor='none')
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
