```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('StudentPerformanceFactors.csv')

df['Study_Category'] = pd.cut(df['Hours_Studied'], bins=[0, 10, 20, 30, 45], labels=['1-10 Hours', '11-20 Hours', '21-30 Hours', '31+ Hours'])

pivot_table = df.pivot_table(index='Sleep_Hours', columns='Study_Category', values='Exam_Score', aggfunc='mean', observed=True)

plt.figure(figsize=(9, 6), dpi=300)

cmap = sns.color_palette("YlGnBu", as_cmap=True)

ax = sns.heatmap(
    pivot_table, 
    annot=True, 
    fmt=".1f", 
    cmap=cmap, 
    linewidths=1.5,
    cbar_kws={'label': 'Average Exam Score'},
    annot_kws={"size": 11, "weight": "bold"}
)

plt.title("Rest & Rigor: The Synergy of Sleep and Study Hours", fontsize=14, fontweight='bold', pad=15, color='#1A365D')
plt.xlabel("Weekly Study Hours Category", fontsize=11, fontweight='bold', color='#1A365D')
plt.ylabel("Daily Sleep Hours", fontsize=11, fontweight='bold', color='#1A365D')

plt.tight_layout()
plt.show()
```
