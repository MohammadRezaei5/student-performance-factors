```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('StudentPerformanceFactors.csv')

df_clean = df.dropna(subset=['Distance_from_Home']).copy()

grouped_tutoring = df_clean.groupby(['Tutoring_Sessions', 'Distance_from_Home'])['Exam_Score'].mean().reset_index()

grouped_tutoring = grouped_tutoring[grouped_tutoring['Tutoring_Sessions'] <= 5]

plt.figure(figsize=(9, 5.5), dpi=300)

palette = {'Near': '#2A9D8F', 'Moderate': '#E9C46A', 'Far': '#E76F51'}

ax = sns.barplot(
    data=grouped_tutoring,
    x='Tutoring_Sessions',
    y='Exam_Score',
    hue='Distance_from_Home',
    hue_order=['Near', 'Moderate', 'Far'],
    palette=palette,
    edgecolor='none',
)

plt.title("Bridging the Gap: Tutoring Support vs. Commute Distance", fontsize=13, fontweight='bold', pad=15, color='#1D3557')
plt.xlabel("Number of Weekly Tutoring Sessions", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylabel("Average Exam Score", fontsize=11, fontweight='bold', color='#1D3557')
plt.ylim(60, 75)

plt.legend(title="Distance from Home", frameon=True, facecolor='white', edgecolor='none')
plt.grid(True, linestyle='--', alpha=0.3, axis='y')
ax.set_axisbelow(True)

plt.tight_layout()
plt.show()
```
